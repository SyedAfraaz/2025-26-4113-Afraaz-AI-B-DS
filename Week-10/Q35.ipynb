```python
import networkx as nx
G = nx.Graph()
G.add_edges_from([
    ('A','B'),
    ('B','C'),
    ('B','D')
])
print(nx.degree_centrality(G))
print(nx.betweenness_centrality(G))
print(nx.closeness_centrality(G))
```
