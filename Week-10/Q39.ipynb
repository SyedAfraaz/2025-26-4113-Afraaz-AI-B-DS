```python
import networkx as nx
G = nx.DiGraph()
G.add_edges_from([
    ('A','B'),
    ('A','c'),
    ('B','C'),
    ('D','C'),
    ('E','B'),
    ('E','C')
])
page_rank = nx.pagerank(G,alpha=0.85)
print("PageRank Scores: ")
for page, rank in page_rank.items():
  print(f"{page}:{rank:4f}")
```
