```python
from sklearn.cluster import KMeans
import numpy as np

X = np.array([
    [2,3],[3,4],[4,2],
    [8,7],[9,6],[10,8]
])

kmeans = KMeans(n_clusters=2, random_state=0)
kmeans.fit(X)

print("Labels:", kmeans.labels_)
print("Centroids:")
print(kmeans.cluster_centers_)
```
